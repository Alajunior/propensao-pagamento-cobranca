# Exploração da Carteira Exemplo

Leitura rápida da base antes de modelar. Toda a lógica está em `src/`; aqui só se olha os dados.

Rode antes: `python dados/gerar_carteira.py`.

```python
import sys; sys.path.insert(0, '..')
import pandas as pd
import matplotlib.pyplot as plt
from src.base_analitica import carregar, montar_base, safras, ALVO

tabelas = carregar()
base = pd.concat(montar_base(d, tabelas) for d in safras())
base.shape
```

## 1. Taxa de pagamento por safra
Base desbalanceada: poucos % pagam em 30 dias. É por isso que acurácia não serve e olhamos AUC, KS e PR-AUC.

```python
base.groupby('data_ref')[ALVO].mean().plot(marker='o', title='Taxa de pagamento em 30 dias por safra');
```

## 2. A população mudou?
A partir de fev/2026 entram clientes com renda menor. É isso que o PSI vai acusar.

```python
base.assign(periodo=base.data_ref >= '2026-02-01').groupby('periodo')['renda_estimada'].describe()
```

## 3. Quem paga?
Taxa de pagamento por faixa de atraso e por produto: os primeiros sinais que o modelo vai aprender.

```python
from src.faixas import faixa_atraso
base.groupby(faixa_atraso(base['dias_atraso']))[ALVO].mean().round(4)
```

```python
base.groupby('produto')[ALVO].agg(['mean', 'size']).round(4)
```
